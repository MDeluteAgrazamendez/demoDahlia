```python
ref = db.reference("students")

while True:
    print("\n===== STUDENT DATABASE MENU =====")
    print("1. Display Students")
    print("2. Add Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Exit")

    choice = input("Enter choice: ")

    # DISPLAY
    if choice == "1":
        students = ref.get()
        print("\nStudent List:")
        for key, student in students.items():
            print(f"ID: {student['id']}, Name: {student['name']}, Math: {student['grades']['Math']}, CS: {student['grades']['CS']}")

    # ADD
    elif choice == "2":
        sid = input("Enter ID: ")
        name = input("Enter name: ")
        math = int(input("Enter Math grade: "))
        cs = int(input("Enter CS grade: "))

        student = {
            "id": int(sid),
            "name": name,
            "grades": {
                "Math": math,
                "CS": cs
            }
        }

        ref.child(sid).set(student)
        print("Student added successfully!")

    # UPDATE
    elif choice == "3":
        sid = input("Enter ID of student to update: ")
        student_ref = ref.child(sid)

        student = student_ref.get()

        if student: # if ref.child(sid).get()
            name = input("Enter new name: ")
            math = int(input("Enter new Math grade: "))
            cs = int(input("Enter new CS grade: "))

            student_ref.update({ #ref.child(sid).update
                "name": name,
                "grades": {
                    "Math": math,
                    "CS": cs
                }
            })

            print("Student updated successfully!")
        else:
            print("Student not found.")

    # DELETE
    elif choice == "4":
        sid = input("Enter ID to delete: ")

        if ref.child(sid).get():
            ref.child(sid).delete()
            print("Student deleted successfully!")
        else:
            print("Student not found.")

    # EXIT
    elif choice == "5":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Try again.")

"""
Loop → keeps menu running

Conditionals (if-elif-else) → handles choices
CRUD operations:
Create → set()
Read → get()
Update → update()
Delete → delete()
Dictionary (JSON) → data structure
Firebase interaction
"""
```
